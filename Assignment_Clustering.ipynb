# Question1: What is unsupervised learning in the context of machine learning?
**Ans:-** Unsupervised learning is a type of machine learning where a model learns patterns from data without labeled answers.

#Question2: How does K-Means clustering algorithm work?
**Ans:-** Working algorithum of Clustering are as follows:-
1. Choose the number of clusters (K)
2. Initialize cluster centroids
3. Assign points to the nearest centroid
4. Update centroids


# Question3: Explain the concept of a dendrogram in hierarchical clustering.
**Ans:-** A dendrogram visually represents the hierarchical structure of clusters, showing when and how data points or groups are merged based on their similarity.

# Question4:  What is the main difference between K-Means and Hierarchical Clustering?
**Ans:-** K-Means requires us to predefine K and produces a single clustering, while hierarchical clustering builds a tree of clusters that offers flexibility in choosing the number of clusters later.

# Question5: What are the advantages of DBSCAN over K-Means?
**Ans:-** Advantages of DBSCAN over K-Means:-
1. DBSCAN automatically discovers the number of clusters based on data density.

   K-Means requires you to choose K in advance.
2. DBSCAN can detect clusters of any shape (e.g., curved or elongated).

   K-Means works best with spherical, evenly sized clusters.    
3. DBSCAN explicitly labels sparse points as noise (outliers).

   K-Means forces every point into a cluster, including outliers.  
4. DBSCAN does not rely on random centroid initialization.

   K-Means can produce different results depending on initial centroids.    
5. DBSCAN is better than K-Means when the data contains noise, outliers, or non-spherical clusters, and when the number of clusters is unknown.


#Question6: When would you use Silhouette Score in clustering?
**Ans:-** we would use the Silhouette Score when we want to evaluate how good our clustering results are, especially to choose or compare the number of clusters.

#Question7: What are the limitations of Hierarchical Clustering?
**Ans:-** Hierarchical clustering is best for small to medium datasets where interpretability matters, but it struggles with scalability, noise, and irreversible early decisions.

#Question8: Why is feature scaling important in clustering algorithms like K-Means?
**Ans:-** Feature scaling is important in clustering algorithms like K-Means because these algorithms rely on distance calculations, and unscaled features can unfairly dominate those distances.

#Question9:  How does DBSCAN identify noise points?

**Ans:-** DBSCAN (Density-Based Spatial Clustering of Applications with Noise) identifies noise points based on local point density, using two key parameters:

ε (epsilon): the radius of the neighborhood around a point.

MinPts: the minimum number of points required to form a dense region.

#Question 10:  Define inertia in the context of K-Means.
**Ans:-** In K-Means clustering, inertia is a measure of how tightly the data points are clustered around their assigned cluster centers.

#Question 11: What is the elbow method in K-Means clustering?
**Ans:-** The elbow method is a heuristic used in K-Means clustering to help choose an appropriate number of clusters k.

#Question 12:  Describe the concept of "density" in DBSCAN.
**Ans:-** In DBSCAN, density is the central concept that determines how clusters are formed.

#Question 13: Can hierarchical clustering be used on categorical data?
**Ans:- Yes, hierarchical clustering can be applied to categorical data, but with some considerations.The key is how distance or similarity between data points is measured.

#Question 14: What does a negative Silhouette Score indicate?
**Ans:-** A negative Silhouette Score indicates that, on average, points are assigned to the wrong clusters.

#Question 15: Explain the term "linkage criteria" in hierarchical clustering.
**Ans:-** In hierarchical clustering, linkage criteria describe how to compute the distance between two clusters from the pairwise distances between their points. This rule then determines which clusters are merged at each step of the algorithm.

#Question 16: Why might K-Means clustering perform poorly on data with varying cluster sizes or densities?
**Ans:-** K-Means tends to perform poorly when clusters have very different sizes or densities because its mathematical assumptions implicitly favor equal-sized, equally dense, roughly spherical clusters.

#Question 17: What are the core parameters in DBSCAN, and how do they influence clustering?
**Ans:-** DBSCAN has two core parameters: epsilon (ε) and MinPts. Together they define what “dense region” means and therefore control which points form clusters and which are treated as noise.

#Question 18:- How does K-Means++ improve upon standard K-Means initialization?
**Ans:-** K-Means++ is an improved version of standard K-Means that addresses a major problem: sensitive initialization of centroids.

#Question 19: What is agglomerative clustering?
**Ans:-** Agglomerative clustering is a type of hierarchical clustering that builds clusters in a bottom-up manner.

#Question20:- What makes Silhouette Score a better metric than just inertia for model evaluation?
**Ans:-** Silhouette Score is often more informative than inertia because it evaluates both cluster compactness and cluster separation, while inertia only measures compactness and always decreases as the number of clusters increases. This makes Silhouette Score more useful for comparing models with different numbers of clusters and for detecting overfitting in clustering.



```python
#Question21
'''
 Generate synthetic data with 4 centers using make_blobs and apply K-Means clustering. Visualize using a
scatter plot
'''
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Generate synthetic data with 4 centers
X, y_true = make_blobs(
    n_samples=500,
    centers=4,
    cluster_std=0.60,
    random_state=42)

# Apply K-Means clustering
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10)
labels = kmeans.fit_predict(X)

#  Visualize using a scatter plot
plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1], c=labels, s=20, cmap='viridis')
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    c='red',
    s=100,
    marker='X',
    label='Centers')
plt.title('K-Means on make_blobs Data with 4 Centers')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

```python
#Question 29;
'''
 Create synthetic data using make_blobs and evaluate silhouette scores
 for k = 2 to 5. Display as a bar chart
 '''
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# 1. Generate synthetic data
X, y_true = make_blobs(
    n_samples=500,
    centers=4,
    cluster_std=0.7,
    random_state=42)

# 2. Evaluate silhouette scores for k = 2 to 5
ks = []
scores = []
for k in range(2, 6):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X)
    score = silhouette_score(X, labels)
    ks.append(k)
    scores.append(score)

# 3. Plot as bar chart
plt.figure(figsize=(6, 4))
plt.bar(ks, scores, color='skyblue')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Silhouette score')
plt.title('Silhouette scores for k = 2 to 5')
plt.xticks(ks)
plt.ylim(0, 1)
plt.show()
```

```python
#Question 30:
'''
Load the Iris dataset and use hierarchical clustering to group data.
Plot a dendrogram with average linkage.
'''
from sklearn.datasets import load_iris
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt

# 1. Load Iris dataset
iris = load_iris()
X = iris.data

# 2. Perform hierarchical clustering with average linkage
Z = linkage(X, method='average')

# 3. Plot dendrogram
plt.figure(figsize=(10, 5))
dendrogram(
    Z,
    truncate_mode=None,
    leaf_rotation=90,
    leaf_font_size=8)
plt.title('Hierarchical Clustering Dendrogram (Average linkage) - Iris')
plt.xlabel('Sample index')
plt.ylabel('Distance')
plt.tight_layout()
plt.show()
```

```python
#Question 31:
'''
Generate synthetic data with overlapping clusters using make_blobs, then apply K-Means and visualize with decision boundaries
'''

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic data with overlapping clusters
X, y_true = make_blobs(
    n_samples=500,
    centers=3,
    cluster_std=2.0,   # large std → more overlap
    random_state=42)

# Fit K-Means
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)
# Create a grid over feature space for decision boundaries
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 300),
    np.linspace(y_min, y_max, 300))

grid_points = np.c_[xx.ravel(), yy.ravel()]
Z = kmeans.predict(grid_points)
Z = Z.reshape(xx.shape)
plt.figure(figsize=(7, 5))

# Decision regions
plt.contourf(xx, yy, Z, alpha=0.3, cmap='Pastel1')

# Clustered data points
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    s=20,
    cmap='tab10',
    edgecolor='k')

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    c='red',
    s=120,
    marker='X',
    label='Centroids')

plt.title('K-Means on Overlapping Blobs with Decision Boundaries')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#Question32:
'''
 Load the Digits dataset and apply DBSCAN after reducing dimensions with t-SNE.
 Visualize the results.
 '''
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
import numpy as np

# Load Digits dataset
digits = load_digits()
X = digits.data

# Reduce dimensions with t-SNE to 2D
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
X_2d = tsne.fit_transform(X)

# Apply DBSCAN on the 2D t-SNE embedding
dbscan = DBSCAN(eps=8, min_samples=5)
labels = dbscan.fit_predict(X_2d)

# Visualize results
plt.figure(figsize=(8, 6))
colors = ['red' if label == -1 else plt.cm.tab10(label) for label in labels]

plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=colors,
    s=15,
    alpha=0.7)

plt.title('DBSCAN on Digits Dataset (t-SNE 2D embedding)')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.colorbar(label='Cluster label (red=outliers)')
plt.tight_layout()
plt.show()
```

```python
#Question34:
'''
 Load the Breast Cancer dataset and compare inertia values for K = 2 to 6 using K-Means.
 Show results in a line plot.
'''
from sklearn.datasets import load_breast_cancer
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

#  Load Breast Cancer dataset
cancer = load_breast_cancer()
X = cancer.data

#  Compute inertia for K = 2 to 6
ks = range(2, 7)
inertias = []
for k in ks:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    inertias.append(kmeans.inertia_)

#  Plot elbow curve
plt.figure(figsize=(7, 5))
plt.plot(ks, inertias, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Number of clusters (K)')
plt.ylabel('Inertia')
plt.title('K-Means Elbow Plot: Inertia vs K (Breast Cancer)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#Question41:
'''
Generate synthetic blobs with 5 centers and apply KMeans. Then use silhouette_score to evaluate the clustering
'''
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
from sklearn.metrics import silhouette_samples


# Generate synthetic blobs with 5 centers
X, y_true = make_blobs(
    n_samples=1000,
    centers=5,
    cluster_std=0.8,
    random_state=42)

# Apply K-Means with 5 clusters
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

# Compute silhouette score
sil_score = silhouette_score(X, labels)
print(f"Silhouette Score: {sil_score:.3f}")

# Visualize clustering + silhouette analysis
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Plot clusters
ax1.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=20)
ax1.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
           c='red', s=150, marker='X', label='Centroids')
ax1.set_title(f'K-Means (5 clusters)\nSilhouette = {sil_score:.3f}')
ax1.legend()

# Silhouette plot
silhouette_samples_values = silhouette_samples(X, labels)
silhouette_avg = silhouette_score(X, labels)

y_lower = 10
for i in range(kmeans.n_clusters):
    # Aggregate the silhouette scores for samples belonging to cluster i, and sort them
    ith_cluster_silhouette_values = silhouette_samples_values[labels == i]
    ith_cluster_silhouette_values.sort()

    size_cluster_i = ith_cluster_silhouette_values.shape[0]
    y_upper = y_lower + size_cluster_i

    color = plt.cm.nipy_spectral(float(i) / kmeans.n_clusters)
    ax2.fill_betweenx(
        np.arange(y_lower, y_upper),
        0,
        ith_cluster_silhouette_values,
        facecolor=color,
        edgecolor=color,
        alpha=0.7)
# Label the silhouette plots with their cluster numbers at the middle
ax2.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

# Compute the new y_lower for next plot
y_lower = y_upper + 10 # 10 for the 0 samples

ax2.set_title('Silhouette Plot for each cluster')
ax2.axvline(silhouette_avg, color="red", linestyle="--", label=f"Avg: {silhouette_avg:.3f}")
ax2.set_xlabel("Silhouette coefficient values")
ax2.set_ylabel("Cluster label")
ax2.set_yticks([])
ax2.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
#Question42:
'''
Load the Breast Cancer dataset, reduce dimensionality using PCA, and apply Agglomerative Clustering.
Visualize in 2D
'''

from sklearn.datasets import load_breast_cancer
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

# Load Breast Cancer dataset
cancer = load_breast_cancer()
X = cancer.data

# Reduce dimensionality to 2D using PCA
pca = PCA(n_components=2, random_state=42)
X_2d = pca.fit_transform(X)

# Apply Agglomerative Clustering (2 clusters)
agg = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward')
labels = agg.fit_predict(X_2d)

# Visualize in 2D
plt.figure(figsize=(8, 6))

plt.scatter(
    X_2d[:, 0],
    X_2d[:, 1],
    c=labels,
    cmap='coolwarm',
    s=40,
    edgecolor='k',
    alpha=0.8)

plt.title('Agglomerative Clustering on Breast Cancer (2D PCA)')
plt.xlabel(f'PC 1 ({pca.explained_variance_ratio_[0]:.1%} variance)')
plt.ylabel(f'PC 2 ({pca.explained_variance_ratio_[1]:.1%} variance)')
plt.colorbar(label='Cluster label')

# Explained variance
print(f"Total variance explained by 2 PCs: {pca.explained_variance_ratio_.sum():.1%}")

plt.tight_layout()
plt.show()
```

```python
#Question43:
'''
Generate noisy circular data using make_circles and visualize clustering results from KMeans and DBSCAN side-by-side.
'''

from sklearn.datasets import make_circles
from sklearn.cluster import KMeans, DBSCAN
import matplotlib.pyplot as plt

# Generate noisy circular data
X, y_true = make_circles(
    n_samples=500,
    factor=0.5,
    noise=0.1,
    random_state=42)

# K-Means (2 clusters)
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X)

# DBSCAN (tuned for circles)
dbscan = DBSCAN(eps=0.18, min_samples=8)
dbscan_labels = dbscan.fit_predict(X)

# Side-by-side visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# K-Means result
ax1.scatter(X[:, 0], X[:, 1], c=kmeans_labels, cmap='coolwarm', s=30, edgecolor='k')
ax1.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
           c='red', s=150, marker='X', label='Centroids')
ax1.set_title('K-Means (Poor: splits circles)')
ax1.legend()

# DBSCAN result (outliers in black)
colors_db = ['black' if label == -1 else plt.cm.viridis(label / 2) for label in dbscan_labels]
ax2.scatter(X[:, 0], X[:, 1], c=colors_db, s=30, edgecolor='k')
ax2.set_title('DBSCAN (Better: follows rings)')

# True structure for reference
fig.suptitle('Noisy Circles: K-Means vs DBSCAN', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
#Question44:
'''
Load the Iris dataset and plot the Silhouette Coefficient for each sample after KMeans clustering
'''

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples, silhouette_score
import matplotlib.pyplot as plt
import numpy as np

# Load Iris dataset and apply K-Means
iris = load_iris()
X = iris.data
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

# Compute silhouette coefficients
silhouette_coeffs = silhouette_samples(X, labels)
avg_silhouette = silhouette_score(X, labels)
print(f"Average Silhouette Score: {avg_silhouette:.3f}")

# Plot silhouette analysis
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Left: Clustering result
ax1.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=40)
ax1.set_title('K-Means Clustering (Iris)')
ax1.set_xlabel('Sepal Length')
ax1.set_ylabel('Sepal Width')

# Right: Silhouette plot
y_lower = 10
for i in range(kmeans.n_clusters):
    ith_cluster_silhouette_values = silhouette_coeffs[labels == i]
    ith_cluster_silhouette_values.sort()

    size_cluster_i = ith_cluster_silhouette_values.shape[0]
    y_upper = y_lower + size_cluster_i

    color = plt.cm.nipy_spectral(float(i) / kmeans.n_clusters)
    ax2.fill_betweenx(
        np.arange(y_lower, y_upper),
        0,
        ith_cluster_silhouette_values,
        facecolor=color,
        edgecolor=color,
        alpha=0.7)
    y_lower = y_upper + 10

ax2.axvline(avg_silhouette, color="red", linestyle="--", linewidth=2,
           label=f"Avg: {avg_silhouette:.3f}")
ax2.axhline(0, color='k', linestyle='-', linewidth=1)
ax2.set_xlim([-0.1, 1])
ax2.set_ylim([0, len(X) + (kmeans.n_clusters + 1) * 10])
ax2.set_title('Silhouette Coefficients per Sample')
ax2.set_xlabel('Silhouette Coefficient')
ax2.set_ylabel('Sample Index')
ax2.set_yticks([])
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
#Question46:
'''
Load the Wine dataset, apply KMeans, and visualize the cluster assignments in a
seaborn pairplot (first 4 features)
'''

from sklearn.datasets import load_wine
from sklearn.cluster import KMeans
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load Wine dataset
wine = load_wine()
X = wine.data
feature_names = wine.feature_names[:4]

# Apply K-Means (3 clusters matching wine classes)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

# Create DataFrame
wine_df = pd.DataFrame(X[:, :4], columns=feature_names)
wine_df['cluster'] = labels

# Seaborn pairplot with cluster coloring
plt.figure(figsize=(10, 8))
sns.pairplot(
    wine_df,
    hue='cluster',
    palette='Set1',
    diag_kind='kde',
    plot_kws={'alpha': 0.7, 's': 30})

plt.suptitle('K-Means Clusters on Wine Dataset (First 4 Features)', y=1.02)
plt.show()

```

```python
#Question47:
'''
Generate noisy blobs using make_blobs and use DBSCAN to identify both clusters and noise points.
Print the count.
'''

from sklearn.datasets import make_blobs
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
import numpy as np

# Generate noisy blobs + extra noise
X, y_true = make_blobs(
    n_samples=500,
    centers=3,
    cluster_std=1.5,
    random_state=42)
noise_points = np.random.uniform(-12, 12, (100, 2))
X_noisy = np.vstack([X, noise_points])

# Apply DBSCAN
dbscan = DBSCAN(eps=0.7, min_samples=8)
labels = dbscan.fit_predict(X_noisy)

# Count clusters and noise
unique_labels, counts = np.unique(labels, return_counts=True)
n_clusters = len([l for l in unique_labels if l != -1])
n_noise = counts[unique_labels == -1][0] if -1 in unique_labels else 0

print(f"Number of clusters: {n_clusters}")
print(f"Number of noise points: {n_noise}")
print("Label counts:", dict(zip(unique_labels, counts)))

# Visualize
plt.figure(figsize=(8, 6))
colors = ['red' if l == -1 else plt.cm.viridis(l / max(unique_labels)) for l in labels]
plt.scatter(X_noisy[:, 0], X_noisy[:, 1], c=colors, s=30, alpha=0.7)
plt.title('DBSCAN on Noisy Blobs (Red = Noise)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```
